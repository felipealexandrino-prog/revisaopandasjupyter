```python
import pandas as pd
dados = {
    "jogador": [
        "Everton Ribeiro", "Erick Pulga", "Jean Lucas",
        "Luciano Juba", "Everton Ribeiro", "Erick Pulga",
        "Jean Lucas", "Luciano Juba"
    ],
    "posicao": [
        "Meia", "Atacante", "Meia", "Lateral",
        "Meia", "Atacante", "Meia", "Lateral"
    ],
    "finalizacoes": [2, 5, 3, 1, 4, 6, 2, 3],
    "passes_certos": [45, 18, 52, 38, 41, 20, 48, 35],
    "nota": [7.2, 8.5, 7.8, 7.0, 7.6, 9.0, 7.4, 7.3]
}
dataframe = pd.DataFrame(dados)
dataframe
```
