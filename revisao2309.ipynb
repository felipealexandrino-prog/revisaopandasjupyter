```python
import pandas as pd

dados = {
    "vendedor": [
        "Ana", "Ana", "Ana",
        "Bruno", "Bruno", "Bruno",
        "Carlos", "Carlos", "Carlos", "Ana"
    ],
    
    "cidade": [
        "Salvador", "Lauro de Freitas", "Salvador",
        "Salvador", "Lauro de Freitas", "Salvador",
        "Lauro de Freitas", "Salvador", "Lauro de Freitas",
        "Lauro de Freitas"
    ],
    
    "categoria": [
        "Eletrônicos", "Acessórios", "Eletrônicos",
        "Acessórios", "Eletrônicos", "Eletrônicos",
        "Acessórios", "Eletrônicos", "Acessórios", "Eletrônicos"
    ],
    
    "produto": [
        "Notebook", "Mouse", "Monitor",
        "Teclado", "Notebook", "Monitor",
        "Mouse", "Notebook", "Teclado", "Notebook"
    ],
    
    "total": [
        3500, 150, 1200,
        200, 3200, 1100,
        180, 3000, 250, 3400
    ]
}

dataframe = pd.DataFrame(dados)
dataframe
```

```python
produto_fat_cidade = pd.pivot_table(dataframe,index="produto",values="total",columns="cidade",aggfunc = ("sum"))
produto_fat_cidade
```

```python
cidade_faturamento_quantidade = dataframe.groupby("cidade").agg(total_vendido=("total","sum"),quantidade_total=("produto","count")).reset_index().sort_values(by="total_vendido",ascending=False)
cidade_faturamento_quantidade
```

```python
vendedor_faturamento_categoria = dataframe.groupby(["vendedor","categoria"]).agg(total_vendido=("total","sum")).reset_index().sort_values(by="total_vendido",ascending=False)
vendedor_faturamento_categoria
```
