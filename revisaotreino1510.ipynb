```python

import pandas as pd

dados = {
    "jogador": [
        "Everton Ribeiro", "Jean Lucas", "Erick Pulga",
        "Luciano Juba", "Cauly", "Everton Ribeiro",
        "Jean Lucas", "Erick Pulga", "Cauly",
        "Luciano Juba", "Everton Ribeiro", "Jean Lucas",
        "Cauly", "Erick Pulga", "Luciano Juba"
    ],
    "posicao": [
        "Meia", "Meia", "Atacante", "Lateral", "Meia",
        "Meia", "Meia", "Atacante", "Meia", "Lateral",
        "Meia", "Meia", "Meia", "Atacante", "Lateral"
    ],
    "finalizacoes": [
        2, 1, 5, 1, 3, 4, 2, 6, 2, 2, 3, 4, 1, 4, 3
    ],
    "passes_certos": [
        48, 55, 17, 36, 42, 51, 61, 21, 38, 44,
        46, 53, 49, 19, 39
    ],
    "desarmes": [
        1, 4, 0, 3, 2, 2, 5, 1, 1, 4, 0, 3, 2, 0, 5
    ],
    "nota": [
        7.2, 7.5, 8.4, 7.0, 7.8, 8.1, 8.0, 9.0,
        6.9, 7.6, 7.4, 8.2, 7.1, 8.7, 7.9
    ]
}
dadosjogadores = pd.DataFrame(dados)
dadosjogadores
```

```python
tamanho = len(dadosjogadores)
print(f"The dataframe has {tamanho} registers!")
```

```python
counter = 0
print(f"Players with eight or more points on a match!\n")
for indice,linha in dadosjogadores.iterrows():
    if linha["nota"] >= 8:
        counter += 1
        print(f"Player: {linha['jogador']}\nGrade: {linha['nota']}\n")
print(f"{counter} registers got eight or more points in a single match.")        
```

```python
counterstealpass = 0
for indice,linha in dadosjogadores.iterrows():
    if linha["passes_certos"] > 40 and linha["desarmes"] >= 2:
        counterstealpass += 1
        print(f"Player: {linha['jogador']}\nSteals: {linha['desarmes']}\nCorrect Passes: {linha['passes_certos']}\n")

print(f"{counterstealpass} registers had got more than 40 correct passes and 2+ steals.")
```

```python
sumshots = 0

for indice,linha in dadosjogadores.iterrows():
    if linha["posicao"] == "Atacante":
        sumshots += linha['finalizacoes']

print(f"The strikers had kicked {sumshots} times.")
    
```

```python
topgrade = dadosjogadores["nota"].idxmax()
datatopgrade = dadosjogadores.loc[topgrade]
print(f"The player with the highest grade is: \n{datatopgrade['jogador']} with {datatopgrade['nota']} points and {datatopgrade['passes_certos']} correct passes in a match")
```
