```python
dataframe = dataframe.fillna({
    "cidade" : "S/Cidade",
    "categoria" : "S/Categoria",
    "preco" : 0,
    "total" : 0
})
dataframe
```

```python
dataframe.duplicated()
```

```python
dataframe = dataframe.drop_duplicated()
```

```python
vendasmaior2000 = dataframe[dataframe["total"] > 2000]
vendasmaior2000
```

```python
cidade_salvador = dataframe[dataframe["cidade"] == "Salvador"]
cidade_salvador
```

```python
total_vendido_ssa = cidade_salvador["total"].sum()
total_vendido_ssa
```

```python
dataframe_vendedor = dataframe.groupby("vendedor").agg(
    faturamento_total=("total" , "sum"),
    quantidade_vendida=("quantidade" , "sum")
)

dataframe_vendedor.reset_index()
```

```python
pivot_cidade_vendedor = pd.pivot_table(dataframe,index="vendedor",columns="cidade",values="total",aggfunc="sum")
pivot_cidade_vendedor
```

```python
pivot_categoria_cidade = pd.pivot_table(dataframe,index="categoria",columns="cidade",values="total",aggfunc="sum")
pivot_categoria_cidade
```

```python
maiorvalor = pivot_categoria_cidade["Salvador"].idxmax()
datatop = pivot_categoria_cidade["Salvador"].loc[maiorvalor]
datatop
```
