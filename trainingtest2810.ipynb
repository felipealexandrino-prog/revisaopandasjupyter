```python
import pandas as pd

match_data = {
    "player": [
        "Lucas Ribeiro", "Mateus Santos", "Joao Pedro",
        "Gabriel Costa", "Rafael Lima", "Enzo Martins",
        "Pedro Henrique", "Caio Oliveira", "Bruno Alves",
        "Samuel Rocha", "Vitor Silva", "Davi Souza"
    ],
    "position": [
        "Forward", "Midfielder", "Defender", "Forward",
        "Midfielder", "Defender", "Goalkeeper", "Forward",
        "Defender", "Midfielder", "Forward", "Goalkeeper"
    ],
    "minutes_played": [90, 75, 90, 65, 80, 90, 90, 30, 85, 70, 60, 90],
    "successful_passes": [18, 54, 42, 12, 61, 48, 25, 7, 44, 58, 16, 32],
    "shots_on_target": [5, 2, 0, 7, 1, 1, 0, 4, 0, 2, 6, 0],
    "rating": [8.7, 7.9, 8.1, 9.0, 7.4, 8.3, 8.5, 7.8, 8.2, 7.6, 8.8, 8.0]
}

match_df = pd.DataFrame(match_data)
match_df
```

```python
top5playersrating = match_df.nlargest(5,"rating")
print(top5playersrating[['player','position','rating']])
```

```python
top3playerssucpas = match_df.nlargest(3,"successful_passes")
print(top3playerssucpas[['player','successful_passes']])
```

```python
filtered_df = match_df[(match_df["minutes_played"] >= 80)]
filtered_df
```

```python
top1shots = filtered_df.nlargest(1,"shots_on_target")
print(top1shots[["player","position","shots_on_target"]])
```

```python
filtered70_df = match_df[(match_df["minutes_played"] >= 70 & (match_df["successful_passes"] >= 40))]
filtered70_df
```

```python
totalrecords = len(filtered70_df)
print(filtered70_df[["player","minutes_played","rating"]])
print(f"\n{totalrecords} records have met both conditions!")
```

```python
lowestrating = filtered70_df["rating"].idxmin()
lowestplayer = filtered70_df.loc[lowestrating]
print(lowestplayer[["player","position","rating"]])
```

```python
filtered2shots = match_df[(match_df["shots_on_target"] >= 2)]
print(filtered2shots[["player","position","shots_on_target","rating"]])
averagerating = filtered2shots["rating"].mean()
print(f"\nThe average rating of these players is {averagerating}!")
```

```python
top2playersfilt = filtered2shots.nlargest(2,"shots_on_target")
print(top2playersfilt[["player","position","shots_on_target"]])
```
