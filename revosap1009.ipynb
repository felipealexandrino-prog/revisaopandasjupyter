```python
dataframe = pd.DataFrame(dados)
```

```python
dataframe["preco"] = dataframe["preco"].fillna(0)
```

```python
dataframe["quantidade"] = dataframe["quantidade"].fillna(0)
```

```python
dataframe = dataframe.drop_duplicates()
```

```python
dataframe["cidade"].value_counts()
```

```python
dataframe["cidade"] = dataframe["cidade"].replace("salvador" , "Salvador")
```

```python
dataframe["preco"] = pd.to_numeric(dataframe["preco"])
```

```python
dataframe.info()
```

```python
dataframe
```

```python
dataframe["total"] = dataframe["quantidade"] * dataframe["preco"]
```

```python
relat_vend = dataframe.groupby("vendedor").agg(
    faturamento_total=("total","sum"),
    quantidade_total=("quantidade","sum")
).reset_index().sort_values(by="faturamento_total",ascending=False)
relat_vend
```

```python
relat_cid = dataframe.groupby("cidade").agg(
    faturamento_total=("total","sum"),
    quantidade_total=("quantidade","sum")
).reset_index().sort_values(by="faturamento_total",ascending=False)
relat_cid
```

```python
dataframeprod = dataframe.groupby("produto").agg(
    faturamento_total=("total","sum")
).reset_index()
dataframeprod
```

```python
topprod = dataframeprod["faturamento_total"].idxmax()
dataframetopprod = dataframeprod.loc[topprod]
dataframetopprod
```
