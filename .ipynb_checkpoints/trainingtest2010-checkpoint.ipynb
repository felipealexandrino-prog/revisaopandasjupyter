```python
import pandas as pd

data = {
    "player": [
        "Roger Gabriel", "Dell", "Ryan Nascimento",
        "Pedrinho", "Wendel", "Dyllan",
        "Caio Suassuna", "Saymon", "Emmanuel",
        "Fred Mundoco", "Kauê Furquim", "Roger Gabriel",
        "Dell", "Ryan Nascimento", "Pedrinho"
    ],
    "position": [
        "Midfielder", "Forward", "Forward",
        "Midfielder", "Defensive Midfielder", "Forward",
        "Forward", "Defensive Midfielder", "Midfielder",
        "Forward", "Forward", "Midfielder",
        "Forward", "Forward", "Midfielder"
    ],
    "age": [19, 17, 17, 20, 19, 18, 19, 18, 17, 18, 17, 19, 17, 17, 20],
    "training_attendance": [
        95, 88, 92, 90, 96, 85, 91, 94, 87, 89, 93, 97, 90, 86, 92
    ],
    "goals": [2, 4, 3, 1, 0, 5, 2, 0, 1, 3, 4, 1, 6, 2, 2],
    "rating": [8.2, 8.7, 8.4, 7.8, 7.6, 8.1, 8.0, 7.5, 7.3, 7.9, 8.5, 8.6, 9.1, 7.8, 8.0]
}

academydataframe = pd.DataFrame(data)
academydataframe
```

```python
totalrecords = len(academydataframe)
print(f"{totalrecords} records were registered!")

meanrating = academydataframe["rating"].mean()
print(f"{meanrating} points was the mean of the rating")


highestrating = academydataframe["rating"].idxmax()
highestratingdf = academydataframe.loc[highestrating]
print(f"The highest rating registered was {highestratingdf['rating']}\nData:\n")
highestratingdf
```

```python
counter = 0
print(f"Players with atleast 8.5 rating:\n")
for index,line in academydataframe.iterrows():
    if line["rating"] >= 8.5:
        counter += 1
        print(f"Player: {line['player']} ---- Position: {line['position']} ---- Rating: {line['rating']}")

print(f"{counter} records have atleast 8.5 rating")
```

```python
counterattendance = 0
print(f"Players with attendance below 90% : \n")
for index,line in academydataframe.iterrows():
    if line["training_attendance"] < 90:
        counterattendance += 1
        print(f"Player: {line['player']} ---- Training Attendance {line['training_attendance']}\n")

print(f"{counterattendance} records have a training attendance below 90%")
```

```python
academydataframe["development_status"] = academydataframe["rating"].apply(lambda x: "Excelent" if x >= 8.5 else ("Promising" if x >= 8.0 else "Needs Improvement")
)
academydataframe
```

```python
finalcounter = 0
for index,line in academydataframe.iterrows():
    if line["goals"] >= 3 and line["training_attendance"] >= 90:
        finalcounter += 1
        print(f"Player: {line['player']} ---- Scored Goals: {line['goals']} ---- Training Attendance: {line['training_attendance']} ---- Rating: {line['rating']}")

print(f"{finalcounter} records have atleast 3 scored goals and a Training attendance with atleast 90 percent!")
```

```python
academydataframe["goal_status"] = academydataframe["goals"].apply(lambda x: "Top Scorer" if x >= 4 else("Good Scoring" if x >= 2 and x <= 3 else "Needs more goals"))
academydataframe                                        
```

```python
academydataframe["age_analyze"] = academydataframe["age"].apply(lambda x:"Promise" if x <= 18 else ("Regular" if x < 20 else "Too Old"))
academydataframe
                                                            
```
