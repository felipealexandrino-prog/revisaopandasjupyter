```python
import pandas as pd

data = {
    "player": [
        "Lucas Ribeiro", "Mateus Santos", "Joao Pedro",
        "Gabriel Costa", "Rafael Lima", "Enzo Martins",
        "Pedro Henrique", "Caio Oliveira", "Bruno Alves",
        "Samuel Rocha", "Vitor Silva", "Davi Souza"
    ],
    "position": [
        "Forward", "Midfielder", "Defender",
        "Forward", "Midfielder", "Defender",
        "Goalkeeper", "Forward", "Defender",
        "Midfielder", "Forward", "Goalkeeper"
    ],
    "age": [17, 18, 19, 17, 20, 18, 19, 17, 18, 20, 19, 18],
    "training_attendance": [96, 88, 92, 94, 85, 91, 98, 89, 93, 87, 95, 90],
    "rating": [8.7, 7.9, 8.1, 9.0, 7.4, 8.3, 8.5, 7.8, 8.2, 7.6, 8.8, 8.0],
    "successful_passes": [42, 55, 38, 20, 61, 48, 25, 18, 44, 58, 23, 30],
    "shots_on_target": [5, 2, 0, 7, 1, 1, 0, 4, 0, 2, 6, 0],
    "scouting_cost": [1200, 950, 1100, 1500, 800, 1050, 1300, 900, 1000, 850, 1400, 1250]
}

scouting_df = pd.DataFrame(data)

scouting_df
```

```python

counter = 0
print(f"List of players with atleast 82 rating , atleast 90 of training attendance and 18 or younger:\n")
for index,line in scouting_df.iterrows():
    if line["rating"] >= 8.2 and line["training_attendance"] >= 90 and line["age"] <= 18:
        counter += 1
        print(f"Player: {line['player']} --- Age: {line['age']} --- Rating: {line['rating']} --- Training Attendance: {line['training_attendance']}\n")
print(f"A total of {counter} records follow the scout requirements!")
```

```python
filtered_df = scouting_df[(scouting_df['rating'] >= 8.2) & (scouting_df['age'] <= 18) & (scouting_df['training_attendance'] >= 90)]
filtered_df
```

```python
top_index = filtered_df["successful_passes"].idxmax()
top_index_result = filtered_df.loc[top_index]
print(f"Player who follow the requirements with the most successful passes")
top_index_result
```

```python


filtered2_df = scouting_df[(scouting_df["rating"] >= 8) & (scouting_df["scouting_cost"] < 1100)]
for index,line in scouting_df.iterrows():
    if line["rating"] >= 8 and line["scouting_cost"] < 1100:
        print(f"Player: {line['player']} --- Scouting_Cost: {line['scouting_cost']}\n")


scouting_cost_mean = filtered2_df["scouting_cost"].mean()
print(f"The mean of the scouting cost of these players is {scouting_cost_mean} R$")
```
