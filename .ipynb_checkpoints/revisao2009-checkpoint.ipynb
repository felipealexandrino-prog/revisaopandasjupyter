```python
import pandas as pd

dados = {
    "vendedor": [
        "Ana", "Bruno", "Carlos", "Ana", "Bruno",
        "Carlos", "Ana", "Bruno", "Carlos", "Ana",
        "Bruno", "Carlos", "Ana"
    ],

    "cidade": [
        "Salvador", "Camaçari", "salvador", "Camaçari", "Salvador",
        "Camaçari", "Salvador", "camacari", "Salvador", "Camaçari",
        "Salvador", "Camaçari", "Salvador"
    ],

    "categoria": [
        "Eletrônicos", "Acessórios", "Eletrônicos", "Casa",
        "Acessórios", "Eletrônicos", "Casa", "Acessórios",
        "Eletrônicos", "Casa", "Acessórios", "Eletrônicos",
        "Eletrônicos"
    ],

    "produto": [
        "Notebook", "Mouse", "Monitor", "Cadeira",
        "Teclado", "Notebook", "Mesa", "Mouse",
        "Monitor", "Cadeira", "Teclado", "Notebook",
        "Notebook"
    ],

    "preco": [
        3500, 120, "1200", 850,
        None, 3400, 700, 130,
        1150, 900, 210, 3450,
        3500
    ],

    "quantidade": [
        2, 5, 2, 1,
        3, None, 2, 6,
        2, 1, 4, 2,
        2
    ]
}

dataframe = pd.DataFrame(dados)
dataframe
```

```python
dataframe.info()
```

```python
dataframe.drop_duplicates()
```

```python
dataframe["cidade"].value_counts()
```

```python
dataframe.replace("camacari","Camaçari")
```

```python
dataframe.replace("salvador","Salvador")
```

```python
dataframe["preco"] = dataframe["preco"].fillna(0)
```

```python
dataframe["quantidade"] = dataframe["quantidade"].fillna(0.0)
```

```python
dataframe["preco"] = pd.to_numeric(dataframe["preco"]).astype(float)
dataframe["total"] = dataframe["quantidade"] * dataframe["preco"]
dataframe
```

```python
dataframecategoria = dataframe.groupby("categoria").agg(
    faturamento_total=("total","sum"),
    quantidade_total=("quantidade","sum")
).reset_index().sort_values(by="faturamento_total",ascending=False)
dataframecategoria
```
