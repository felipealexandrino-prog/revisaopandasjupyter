```python
import pandas as pd

players = pd.DataFrame({
    "player": [
        "Everton Ribeiro",
        "ERICK PULGA",
        "Jean Lucas",
        "Luciano Juba",
        "everton ribeiro",
        "Cauly",
        "ERICK PULGA",
        "Jean Lucas",
        "Cauly",
        "Luciano Juba"
    ],
    "position": [
        "Midfielder",
        "Forward",
        "Midfielder",
        "Left Back",
        "MIDFIELDER",
        "Midfielder",
        "FORWARD",
        "MIDFIELDER",
        "MIDFIELDER",
        "LEFT BACK"
    ],
    "competition": [
        "Brasileirao",
        "Copa do Brasil",
        "Brasileirao",
        "Brasileirao",
        "Brasileirao",
        "Copa do Brasil",
        "Brasileirao",
        "Copa do Brasil",
        "Brasileirao",
        "Copa do Brasil"
    ],
    "goals": [1, 3, 0, 1, 2, 1, 2, 1, 0, 0],
    "rating": [7.5, 8.7, 7.2, 7.8, 8.1, 7.6, 9.0, 7.9, 7.1, 7.4]
})

players
```

```python
evertoncount = players["player"].str.contains("Everton",case=False)

evertoncounttop = players.loc[evertoncount]
evertoncounttop
```

```python
jean_appears = players["player"].str.contains("Jean",case=False)
result = jean_appears.sum()
print(f"{result} records has Jean in the player name")
```

```python
players["player_clean"] = players["player"].str.lower()
players
```

```python
search_midfielder = players["position"].str.contains("Midfielder",case=False)
search_midfielderresult = players.loc[search_midfielder]
search_midfielderresult
```

```python
rating = 8
counter = 0 
for index,line in players.iterrows():
    if line['competition'] == "Brasileirao" and line['rating'] >= rating:
        counter += 1

print(f"{counter} records of Brasileirão were rated with {rating} or more points!")
```
