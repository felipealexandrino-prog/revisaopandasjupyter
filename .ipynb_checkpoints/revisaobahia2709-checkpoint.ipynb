```python
import pandas as pd

dados = {
    "jogador": [
        "Everton Ribeiro",
        "Erick Pulga",
        "Jean Lucas",
        "Luciano Juba",
        "Willian José",
        "Nestor",
        "Ademir",
        "Erick",
        "Alejo Veliz",
        "Kike Olivera",
        "Everton Ribeiro",
        "Erick Pulga"
    ],

    "competicao": [
        "Brasileirao",
        "Brasileirao",
        "Brasileirao",
        "Brasileirao",
        "Brasileirao",
        "Copa do Brasil",
        "Copa do Brasil",
        "Copa do Brasil",
        "Copa do Nordeste",
        "Copa do Nordeste",
        "Copa do Nordeste",
        "Copa do Nordeste"
    ],

    "gols": [
        3, 8, 4, 3, 5, 2,
        4, 3, 3, 2, 2, 5
    ],

    "assistencias": [
        5, 4, 3, 6, 2, 4,
        3, 2, 2, 4, 5, 3
    ]
}

dataframe = pd.DataFrame(dados)
dataframe
```

```python
dataframe = dataframe.drop_duplicates()
dataframe
```

```python
dataframe["participacoes"] = dataframe["gols"] + dataframe["assistencias"]
dataframe
```

```python
dataframepart = dataframe.groupby(["jogador","competicao"]).agg(
    participacoes_total = ("participacoes","sum")
).reset_index().sort_values(by="participacoes_total",ascending=False)

dataframepart
```

```python
with pd.ExcelWriter("Dados de desempenho do Bahia.xlsx") as writer:
    dataframe.to_excel(writer,index=False,sheet_name="Dados_Tratados_SemDuplicidade")
    dataframepart.to_excel(writer,index=False,sheet_name="Dados_Desempenho_Por_Competição")
```
